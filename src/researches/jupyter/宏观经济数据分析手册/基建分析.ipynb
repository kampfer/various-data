## 1 基建投资的口径和内涵

### 1.1 广义基建和狭义基建

市场上，广义基建投资是指wind统计的3个主要行业固定资产投资的加总。它与官方认可的统计口径是不同的。

![官方口径和wind口径的基建范围有所不同](./imgs/基建分析_1.png)

官方口径少了仓储业，增加了信息传输、软件和信息技术服务业中的2个子行业

## 2 观测基建投资的三个维度

### 2.1 政府意愿

观察年内重大会议以及关键的政策文件，把握财政政策主基调

1. 两会和政府工作报告（每年）
2. 中共中央政治局会议（每月）
3. 中央经济工作会议（每年）

定量分析：发改委项目批复数据和地方政府重大项目计划投资额

### 2.2 资金来源

基建投资的天花板：本年实际到位资金+上年末结余资金+新增应付款

基建项目资金构成：20%-30%项目资本金 + 70%-80%贷款

基建投资的资金来源：
- 国家预算资金：一般公共预算+政府性预算
- 国内贷款：被低估
- 自筹资金
- 利用外资
- 其他资金

### 2.3 项目建设

从侧面刻画基建投资变动指标：

1. 建筑业上市公司新签订单额。需要从零碎的公告中提取重大项目并汇总，工作量和难度比较大。
2. 建筑业pmi。此指标还受房地产业影响，所以只有房地产投资相对稳定或者方向明确的情况下，才能用。
3. 挖掘机销量。比较滞后
4. 水泥价格指数。
5. 螺纹钢价格
6. 石油沥青
